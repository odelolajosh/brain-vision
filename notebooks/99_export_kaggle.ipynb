```python
# ── CELL 1: Imports and paths ─────────────────────────────────────────────────
import zipfile
from pathlib import Path
from datetime import datetime

CHECKPOINTS_DIR = Path("../checkpoints")
RESULTS_DIR     = Path("../results")
EXPORT_DIR      = Path("../exports")
EXPORT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# ── CELL 2: Bundle and download ───────────────────────────────────────────────
timestamp = datetime.now().strftime("%Y%m%d_%H%M")
zip_path  = EXPORT_DIR / f"brainvision_{timestamp}.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zf:

    # Checkpoints (.pt)
    for f in sorted(CHECKPOINTS_DIR.glob("*.pt")):
        zf.write(f, f"checkpoints/{f.name}")

    # Histories (.npy)
    for f in sorted(RESULTS_DIR.glob("*_history.npy")):
        zf.write(f, f"histories/{f.name}")

size_mb = zip_path.stat().st_size / 1024 / 1024
n_ckpts = sum(1 for _ in CHECKPOINTS_DIR.glob("*.pt"))
n_hist  = sum(1 for _ in RESULTS_DIR.glob("*_history.npy"))

print(f"✅ {zip_path.name}  ({size_mb:.1f} MB)")
print(f"   checkpoints/ : {n_ckpts} files")
print(f"   histories/   : {n_hist} files")
```
