## Visualize data

```python
from brainvision.data import load_patient

patient_004_02 = load_patient("../datasets/first_campaign/004-02.zip")
patient_034_01 = load_patient("../datasets/second_campaign/034-01.zip")
```

```python
import matplotlib.pyplot as plt
from brainvision.visualisation import (
    show_hypercube, show_ground_truth, show_patient, make_pseudo_rgb
)

# ── Standalone RGB ────────────────────────────────────────────────────────────
show_hypercube(patient_004_02['raw'], title=patient_004_02['id'])
plt.show()

# ── RGB with GT overlay ───────────────────────────────────────────────────────
show_hypercube(
    patient_004_02['raw'],
    label_map=patient_004_02['labels'],
    alpha=0.4,
    title=f"{patient_004_02['id']} — GT overlay",
)
plt.show()

# ── Ground truth standalone ───────────────────────────────────────────────────
show_ground_truth(patient_004_02['labels'], title=patient_004_02['id'])
plt.show()

# ── GT with faint RGB background ─────────────────────────────────────────────
rgb = make_pseudo_rgb(patient_004_02['raw'])
show_ground_truth(
    patient_004_02['labels'],
    background_image=rgb,
    alpha=0.6,
    title=patient_004_02['id'],
)
plt.show()

# ── Full three-panel convenience view ────────────────────────────────────────
fig, axes = show_patient(
    patient_004_02['raw'],
    patient_004_02['labels'],
    patient_id=patient_004_02['id'],
)
plt.show()

# # ── Embed into a larger figure ────────────────────────────────────────────────
# fig, axes = plt.subplots(2, 3, figsize=(15, 10))
# for i, patient_004_02 in enumerate(patient_004_02s[:6]):
#     show_hypercube(
#         patient_004_02['processed'],
#         label_map=patient_004_02['labels'],
#         title=patient_004_02['id'],
#         ax=axes.flatten()[i],
#     )
# plt.tight_layout()
# plt.show()
```
